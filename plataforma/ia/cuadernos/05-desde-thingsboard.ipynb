# 5 · Datos desde ThingsBoard

Si ya tienes histórico en ThingsBoard (por ejemplo, de la **serie 2**), no hace falta volver a recogerlo:
`desde_thingsboard` lo descarga por su API REST (una columna por clave).

```python
from labia import desde_thingsboard

tb = desde_thingsboard("Aerogenerador-Aula", ["viento_mps", "potencia_kw", "vibracion_mms"], horas=2)
print(len(tb), "filas")
tb.head()
```

```python
tb.plot(x="viento_mps", y="potencia_kw", kind="scatter", s=5, alpha=.4, figsize=(8, 4), grid=True);
```

Para usarlo como cualquier otro conjunto, guárdalo en `datos/`:

```python
from labia import DATOS
tb.to_csv(DATOS / "aerolab-thingsboard.csv", index=False)
```

> Ojo: ThingsBoard solo guarda lo que se le envió. Las **etiquetas** (escenario) normalmente no están: con estos
> datos solo se pueden entrenar modelos sin etiqueta (anomalías) o de regresión.
